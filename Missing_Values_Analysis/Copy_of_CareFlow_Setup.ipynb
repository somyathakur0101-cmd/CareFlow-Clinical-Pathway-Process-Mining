```python

import pandas as pd
from google.colab import files

uploaded = files.upload()

filename = next(iter(uploaded))

df = pd.read_csv(filename)

print("DATASET LOADED SUCCESSFULLY")
print("-" * 50)

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nFIRST FIVE RECORDS")
display(df.head())

print("\nCOLUMN NAMES")
for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")

print("\nDATA TYPES")
print(df.dtypes)

schema = pd.DataFrame({
    "Column_Name": df.columns,
    "Data_Type": [str(df[col].dtype) for col in df.columns],
    "Non_Missing_Count": df.notna().sum().values,
    "Missing_Count": df.isna().sum().values,
    "Missing_Percentage": (
        df.isna().mean().values * 100
    ).round(2)
})

print("\nDATASET SCHEMA REPORT")
display(schema)

print("\nDATA TYPE SUMMARY")
print(df.dtypes.value_counts())

datetime_columns = [
    col for col in df.columns
    if "datetime" in col.lower()
    or "date" in col.lower()
    or "time" in col.lower()
]

print("\nCOLUMNS THAT MAY CONTAIN DATES/TIMES")
for col in datetime_columns:
    print(f"{col}: current type = {df[col].dtype}")

empty_columns = df.columns[df.isna().all()].tolist()

print("\nCOMPLETELY EMPTY COLUMNS")
if empty_columns:
    print(empty_columns)
else:
    print("No completely empty columns found.")

print("\nNUMERICAL COLUMN SUMMARY")
display(df.describe())

schema.to_csv("CareFlow_Dataset_Schema.csv", index=False)

print("\nSchema report saved successfully.")

files.download("CareFlow_Dataset_Schema.csv")
```


##  Missing Values Analysis

**Objective:** To identify missing values in each column of the original HospitalOPD dataset and calculate their missing percentages.


```python

import pandas as pd
from google.colab import files
from io import BytesIO

uploaded = files.upload()
filename = next(iter(uploaded))

df = pd.read_csv(BytesIO(uploaded[filename]))

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python

df.isnull().sum()
```

```python

(df.isnull().mean() * 100).round(2)
```

```python

missing_report = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing Percentage (%)': (df.isnull().mean() * 100).round(2)
})

display(missing_report)
```

```python
missing_report.to_csv('Member2_Missing_Values_Report.csv')
files.download('Member2_Missing_Values_Report.csv')
```

### Findings – Missing Values Analysis

- Total rows: 62,422
- Total columns: 25
- `Unnamed: 2` has 100% missing values.
- Lab-related columns have approximately 71.12% missing values.
- Imaging-related columns have approximately 91.15%–91.17% missing values.
- Pharmacy-related columns have approximately 99.52% missing values.
- Most appointment and prescription columns have no missing values or very few missing values.
