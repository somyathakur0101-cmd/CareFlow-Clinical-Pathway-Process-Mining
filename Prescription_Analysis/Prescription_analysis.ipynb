```python
# Import Pandas for data handling and analysis
import pandas as pd

# Import NumPy for numerical calculations
import numpy as np

# Import Matplotlib for creating charts
import matplotlib.pyplot as plt

# Import OS to check uploaded files
import os

# Display success message
print("All libraries imported successfully!")
```

```python

from google.colab import files

uploaded = files.upload()
```

```python
# Get the exact uploaded file name
file_name = list(uploaded.keys())[0]

# Read Excel dataset using Pandas
df = pd.read_excel(file_name)

# Display successful loading message
print("Dataset loaded successfully!")

# Display total rows and columns
print("Dataset Shape:", df.shape)

# Display first five rows
df.head()
```

```python
# Display all column names
print("Dataset Columns:")
print(df.columns.tolist())

# Display number of rows and columns
print("\nDataset Information:")
print(df.info())

# Display statistical summary of numerical columns
print("\nStatistical Summary:")
display(df.describe())

# Check missing values in each column
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Define columns required for prescription analysis
required_columns = [
    "prescription_order_id",
    "prescription_doctor_name",
    "prescription_start_datetime",
    "prescription_end_datetime",
    "appointment_start_datetime"
]

# Check whether each required column exists
for col in required_columns:

    if col in df.columns:
        print(col, " - Available")

    else:
        print(col, " - Missing")
```

```python
# Inspect the sheet names available in the uploaded Excel file
xls = pd.ExcelFile(file_name)
print("Available sheets in Excel file:", xls.sheet_names)

# Note: To resolve this, you need to read the correct sheet.
# For example, if there is a 'Clean Data' sheet, you would run:
# df = pd.read_excel(file_name, sheet_name='Clean Data')
```

```python
# Define date and time columns
date_cols = [
    "prescription_start_datetime",
    "prescription_end_datetime",
    "appointment_start_datetime"
]

# Convert columns into datetime format
# errors='coerce' converts invalid dates into NaT
for col in date_cols:

    df[col] = pd.to_datetime(
        df[col],
        errors="coerce"
    )

# Display conversion message
print("Date columns converted successfully!")

# Display converted date columns
display(df[date_cols].head())
```

```python
# Count total rows in the original dataset
print("Total Rows:", len(df))

# Count duplicate prescription IDs
duplicate_count = df.duplicated(
    subset=["prescription_order_id"]
).sum()

print("Duplicate Prescription IDs:", duplicate_count)

# Count unique prescription IDs
unique_count = df["prescription_order_id"].nunique()

print("Unique Prescription IDs:", unique_count)
```

```python
# Keep one representative row for each prescription ID
# This avoids counting the same prescription multiple times
prescription_df = df.drop_duplicates(
    subset=["prescription_order_id"]
).copy()

# Display prescription-level record count
print("Total Prescription Records:", len(prescription_df))

# Display first five records
display(prescription_df.head())
```

```python
# Calculate the difference between prescription end and start time
# Convert the difference from seconds into minutes

prescription_df["prescription_duration"] = (
    prescription_df["prescription_end_datetime"] -
    prescription_df["prescription_start_datetime"]
).dt.total_seconds() / 60

# Display duration statistics
print("Prescription Duration Statistics:")

display(
    prescription_df["prescription_duration"].describe()
)
```

```python
# Calculate the time between appointment start
# and prescription start

prescription_df["waiting_time"] = (
    prescription_df["prescription_start_datetime"] -
    prescription_df["appointment_start_datetime"]
).dt.total_seconds() / 60

# Display waiting time statistics
print("Appointment Waiting Time Statistics:")

display(
    prescription_df["waiting_time"].describe()
)
```

```python
# Calculate total unique prescriptions
total_prescriptions = (
    prescription_df["prescription_order_id"].nunique()
)

# Calculate total unique doctors
total_doctors = (
    prescription_df["prescription_doctor_name"].nunique()
)

# Calculate median prescription duration
median_duration = (
    prescription_df["prescription_duration"].median()
)

# Calculate median waiting time using non-negative values
valid_waiting = prescription_df.loc[
    prescription_df["waiting_time"] >= 0,
    "waiting_time"
].dropna()

median_waiting = valid_waiting.median()

# Display all important metrics
print("========== PRESCRIPTION SUMMARY ==========")

print("Total Prescriptions:", total_prescriptions)

print("Total Doctors:", total_doctors)

print("Median Prescription Duration:", median_duration, "minutes")

print("Median Waiting Time:", median_waiting, "minutes")
```

```python
# Group prescription records by doctor name
# Count unique prescription IDs for each doctor
# Sort doctors in descending order

doctor_count = (
    prescription_df.groupby("prescription_doctor_name")
    ["prescription_order_id"]
    .nunique()
    .sort_values(ascending=False)
)

# Select top 10 doctors
top_10_doctors = doctor_count.head(10)

# Display top 10 doctors
print("Top 10 Doctors by Prescription Count:")

display(top_10_doctors)
```

```python
# Create figure for the bar chart
plt.figure(figsize=(10, 6))

# Plot top 10 doctors in horizontal bar chart
top_10_doctors.sort_values().plot(kind="barh")

# Add chart title and axis labels
plt.title("Top 10 Doctors by Prescription Count")
plt.xlabel("Number of Prescriptions")
plt.ylabel("Doctor Name")

# Adjust chart layout
plt.tight_layout()

# Save chart as PNG image
plt.savefig("01_top_doctors.png")

# Display chart
plt.show()
```

```python
# Extract month and year from prescription start datetime
# Period format groups records month-wise

prescription_df["month"] = (
    prescription_df["prescription_start_datetime"]
    .dt.to_period("M")
)

# Group data month-wise and count unique prescriptions
monthly = (
    prescription_df.groupby("month")
    ["prescription_order_id"]
    .nunique()
)

# Display monthly prescription count
print("Monthly Prescription Count:")

display(monthly)
```

```python
# Create figure for monthly trend chart
plt.figure(figsize=(12, 5))

# Plot monthly prescription counts
monthly.plot(marker="o")

# Add chart title and labels
plt.title("Monthly Prescription Trend")
plt.xlabel("Month")
plt.ylabel("Number of Prescriptions")

# Rotate month labels for better readability
plt.xticks(rotation=45)

# Add grid lines
plt.grid(True)

# Adjust layout
plt.tight_layout()

# Save chart
plt.savefig("02_monthly_trend.png")

# Display chart
plt.show()
```

```python
# Select prescription duration column
duration = prescription_df["prescription_duration"].dropna()

# Count zero-duration records
zero_duration = (duration == 0).sum()

# Count negative-duration records
negative_duration = (duration < 0).sum()

# Count missing duration records
missing_duration = (
    prescription_df["prescription_duration"].isna().sum()
)

# Select only positive duration values
valid_duration = duration[duration > 0]

# Calculate percentage of zero-duration records
zero_percentage = (
    (duration == 0).sum() / len(duration)
) * 100

# Display quality statistics
print("Zero Duration:", zero_duration)

print("Negative Duration:", negative_duration)

print("Missing Duration:", missing_duration)

print("Zero Duration Percentage:", round(zero_percentage, 2), "%")

# Display positive duration statistics
display(valid_duration.describe())
```

```python
# Create figure for duration histogram
plt.figure(figsize=(9, 5))

# Use positive duration records only
# Limit extreme values to 99th percentile for readability

if len(valid_duration) > 0:

    upper_limit = valid_duration.quantile(0.99)

    filtered_duration = valid_duration[
        valid_duration <= upper_limit
    ]

    # Plot histogram
    filtered_duration.hist(bins=30)

# Add title and axis labels
plt.title("Prescription Processing Duration")
plt.xlabel("Duration in Minutes")
plt.ylabel("Frequency")

# Adjust layout
plt.tight_layout()

# Save chart
plt.savefig("03_prescription_duration.png")

# Display chart
plt.show()
```

```python
# Select non-negative waiting time records
valid_waiting = prescription_df.loc[
    prescription_df["waiting_time"] >= 0,
    "waiting_time"
].dropna()

# Display waiting time statistics
print("Waiting Time Statistics:")

display(valid_waiting.describe())

# Calculate median waiting time
print(
    "Median Waiting Time:",
    valid_waiting.median(),
    "minutes"
)
```

```python
# Create figure for waiting time histogram
plt.figure(figsize=(9, 5))

# Remove extreme values for better chart readability
if len(valid_waiting) > 0:

    upper_limit = valid_waiting.quantile(0.99)

    filtered_waiting = valid_waiting[
        valid_waiting <= upper_limit
    ]

    # Plot histogram
    filtered_waiting.hist(bins=30)

# Add chart title and labels
plt.title("Appointment to Prescription Waiting Time")
plt.xlabel("Waiting Time in Minutes")
plt.ylabel("Frequency")

# Adjust layout
plt.tight_layout()

# Save chart
plt.savefig("04_waiting_time.png")

# Display chart
plt.show()
```

```python
# Create a summary table containing important prescription metrics

summary = pd.DataFrame({

    "Metric": [
        "Total Prescriptions",
        "Total Doctors",
        "Median Prescription Duration",
        "Median Waiting Time",
        "Zero Duration Records",
        "Negative Duration Records",
        "Missing Duration Records"
    ],

    "Value": [
        total_prescriptions,
        total_doctors,
        median_duration,
        median_waiting,
        zero_duration,
        negative_duration,
        missing_duration
    ]

})

# Display summary table
display(summary)

# Save summary table as CSV file
summary.to_csv(
    "prescription_summary.csv",
    index=False
)

print("Prescription summary saved successfully!")
```

```python
# Convert doctor-wise series into a DataFrame
doctor_table = doctor_count.reset_index()

# Rename columns for better readability
doctor_table.columns = [
    "Doctor Name",
    "Prescription Count"
]

# Display doctor-wise table
display(doctor_table.head(10))

# Save doctor-wise analysis into CSV
doctor_table.to_csv(
    "doctor_wise_prescription_count.csv",
    index=False
)

print("Doctor-wise report saved successfully!")
```

```python
# Count records where start and end timestamps are identical
identical_timestamps = (
    prescription_df["prescription_start_datetime"] ==
    prescription_df["prescription_end_datetime"]
).sum()

# Create data quality report
quality_report = pd.DataFrame({

    "Metric": [
        "Total Prescription Records",
        "Missing Duration",
        "Zero Duration",
        "Zero Duration Percentage",
        "Negative Duration",
        "Zero Duration with Identical Timestamps"
    ],

    "Value": [
        len(prescription_df),
        missing_duration,
        zero_duration,
        round(zero_percentage, 2),
        negative_duration,
        identical_timestamps
    ]

})

# Display quality report
display(quality_report)

# Save quality report as CSV
quality_report.to_csv(
    "prescription_quality_report.csv",
    index=False
)

print("Quality report saved successfully!")
```
